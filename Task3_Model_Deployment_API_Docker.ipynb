# FastAPI Application

FastAPI is used to create a REST API for model inference.

The API contains two endpoints:

## GET /

Used to check whether the server is running.

## POST /predict

Accepts an image file and returns:

- Predicted class
- Confidence score

The image is resized and normalized before passing it to the deep learning model.

```python
app_code = """

from fastapi import FastAPI, UploadFile, File
from tensorflow.keras.models import load_model
from PIL import Image
import numpy as np
import io


app = FastAPI(
    title="CIFAR-10 Image Classification API"
)


model = load_model(
    "model/cifar10_mobilenet_model.keras"
)


class_names = [
    "Airplane",
    "Automobile",
    "Bird",
    "Cat",
    "Deer",
    "Dog",
    "Frog",
    "Horse",
    "Ship",
    "Truck"
]


@app.get("/")
def home():

    return {
        "message": "API is running successfully"
    }



@app.post("/predict")
async def predict(file: UploadFile = File(...)):

    image_bytes = await file.read()

    image = Image.open(
        io.BytesIO(image_bytes)
    )

    image = image.convert("RGB")

    image = image.resize(
        (96,96)
    )


    image = np.array(image)

    image = image / 255.0


    image = np.expand_dims(
        image,
        axis=0
    )


    prediction = model.predict(image)


    index = np.argmax(prediction)


    confidence = float(
        np.max(prediction)
    )


    return {

        "prediction": class_names[index],

        "confidence": confidence

    }

"""


with open(
    "Task3_Model_Deployment/app.py",
    "w"
) as file:

    file.write(app_code)


print("app.py created successfully")
```

# Docker Configuration

Docker is used to package the application with all required dependencies.

The Docker container:

1. Creates a Python environment
2. Installs required libraries
3. Copies project files
4. Starts the FastAPI server

```python
dockerfile = """

FROM python:3.10


WORKDIR /app


COPY requirements.txt .


RUN pip install --no-cache-dir -r requirements.txt


COPY . .


EXPOSE 8000


CMD ["uvicorn",
"app:app",
"--host",
"0.0.0.0",
"--port",
"8000"]

"""


with open(
    "Task3_Model_Deployment/Dockerfile",
    "w"
) as file:

    file.write(dockerfile)


print("Dockerfile created successfully")
```

# API Testing

The prediction endpoint accepts an image file.

## Endpoint

POST /predict


## Example Request

Upload an image file:

## Example Response

The response contains the predicted class and confidence value.

```python
sample_request = """

import requests


url = "http://127.0.0.1:8000/predict"


files = {

    "file": open(
        "sample_image.png",
        "rb"
    )

}


response = requests.post(
    url,
    files=files
)


print(response.json())

"""


with open(
    "Task3_Model_Deployment/sample_request.py",
    "w"
) as file:

    file.write(sample_request)


print("sample_request.py created successfully")
```
